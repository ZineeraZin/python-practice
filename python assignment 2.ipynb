# 2. Dictionary

```python
student_marks = {
    "Anu": 75,
    "Asha": 88,
    "Diya": 92,
    "Meera": 68,
    "Sara": 81
}

print(student_marks)
```

```python
print("Diya's mark:", student_marks["Diya"])
```

```python
student_marks["Janani"] = 80

print(student_marks)
```

```python
student_marks["Anu"] = 82

print(student_marks)
```

```python
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
```

# 3. Sets

```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print(my_set)
```

```python
my_set[4] = 's'
```

### Explanation:

A set does not allow duplicate values.
Sets are also unordered, so the order of elements may be different when
the code is executed.

```python
set1 = {1, 3, 5, 7, 9}

set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)
print("Set 2:", set2)
```

```python
union_set = set1.union(set2)
intersection_set = set1.intersection(set2)

print("Union:", union_set)
print("Intersection:", intersection_set)
```
