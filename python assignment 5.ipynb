#  Assignment 5: Data Analysis using NumPy and Pandas

## 1. Create a 1D NumPy Array

```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

## 2. Inspection and Properties

```python
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

## 3. Array Operations

```python
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_f)
```

### Find the Maximum, Minimum, and Mean Temperatures

```python
print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

## 4. Array Slicing and Indexing

```python
first_three_days = temperatures_w1[:3]

print("First Three Days:")
print(first_three_days)
```

### Weekend Temperatures

```python
weekend = temperatures_w1[-2:]

print("Weekend Temperatures:")
print(weekend)
```

### Middle Three Days

```python
middle_three_days = temperatures_w1[2:5]

print("Middle Three Days:")
print(middle_three_days)
```

## 5. Create a 2D Array

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print(temperatures)
```

## 6. Inspect and Slice the 2D Array

```python
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)
```

### Extract Temperatures for Each Week

```python
week1 = temperatures[0]
week2 = temperatures[1]

week1_weekend = temperatures[0, -2:]
week2_weekend = temperatures[1, -2:]

print("Week 1:", week1)
print("Week 2:", week2)

print("Week 1 Weekend:", week1_weekend)
print("Week 2 Weekend:", week2_weekend)
```

# Pandas Series

## 1. Creating Pandas Series

```python
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

## 2. Indexing and Slicing

```python
print("Mark of 1st Rank Student:", marks.iloc[0])
```

### Retrieve the Top 3 Ranks

```python
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
```

### Retrieve the 3rd Rank Student

```python
print("Mark of 3rd Rank Student:", marks.iloc[2])
```

### Boolean Mask

```python
print(marks[marks > 90])
```

## 3. Manipulating Series

```python
marks.loc['Rank1'] = 100

print(marks)
```

### Remove the Last Rank

```python
marks = marks.drop('Rank5')

print(marks)
```

### Compute CGPA

```python
cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

# Pandas DataFrame

## 1. Creating Pandas DataFrame

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

transactions
```

## 2. Data Exploration

```python
print("Transactions DataFrame:")
display(transactions)

print("First 5 Rows:")
display(transactions.head())

print("Last 5 Rows:")
display(transactions.tail())

print("Shape:", transactions.shape)

print("Column Names:")
print(transactions.columns)

print("Data Types:")
print(transactions.dtypes)
```

### Display ProductCategory and Amount

```python
display(transactions[['ProductCategory', 'Amount']])
```

### Retrieve the Last 3 Columns


```python
display(transactions.iloc[:, -3:])
```

### Filter Rows

```python
filtered_transactions = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

display(filtered_transactions)
```

### Value Counts

```python
print(transactions['ProductCategory'].value_counts())
```

### Unique Regions

```python
print(transactions['Region'].unique())
```

### Mean Amount by Region

```python
print(transactions.groupby('Region')['Amount'].mean())
```

## 3. Manipulating the DataFrame


```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

display(transactions)
```

### Add Discount Column

```python
transactions['Discount'] = transactions['Amount'] * 0.10

display(transactions)
```

### Remove Transaction 109

```python
transactions = transactions[
    transactions['TransactionID'] != 109
]

display(transactions)
```
